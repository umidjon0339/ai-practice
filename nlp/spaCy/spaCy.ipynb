## Phase 1: Python & NLP Basics
**Concepts**: Python, strings, lists, dictionaries, NLP pipeline, text processing

```python
# Phase 1: Python & NLP Basics
# Before diving deep into spaCy, let's review basic Python string manipulations.

text = "Natural Language Processing (NLP) is a subfield of artificial intelligence."

# Example 1: String methods
print("Lowercase:", text.lower())
print("Uppercase:", text.upper())
print("Split into words:", text.split())

# Example 2: List Comprehensions for basic filtering
words = text.split()
long_words = [word for word in words if len(word) > 5]
print("Words longer than 5 chars:", long_words)

# Example 3: Dictionary for simple frequency counts
word_counts = {}
for word in text.lower().split():
    word_counts[word] = word_counts.get(word, 0) + 1
print("Word Counts:", word_counts)
```

## Phase 2: spaCy Setup
**Concepts**: Installation, language models, pipeline, nlp() object, Doc, Token, Span

```python
# Phase 2: spaCy Setup
# Note: Uncomment the line below to install spacy and the english model if you haven't already.
# !pip install spacy
# !python -m spacy download en_core_web_sm

import spacy

# Load the small English pipeline
# 'en_core_web_sm' is a small, general-purpose model trained on web text.
nlp = spacy.load("en_core_web_sm")

# The nlp object is used to create documents with linguistic annotations.
doc = nlp("Apple is looking at buying U.K. startup for $1 billion.")

# Example 1: Exploring Tokens
# A Token is an individual word or punctuation mark in the text.
print("Tokens:")
for token in doc:
    print(token.text)

# Example 2: Exploring Spans
# A Span is a slice from the document.
span = doc[1:4]
print("\nSpan:", span.text)

# Example 3: The Doc object
print("\nDoc length:", len(doc))
```

## Phase 3: spaCy Architecture
**Concepts**: Pipeline components, vocabulary, lexical attributes, annotations

```python
# Phase 3: spaCy Architecture
import spacy
nlp = spacy.load("en_core_web_sm")

# The processing pipeline consists of multiple components (e.g., tagger, parser, ner).
print("Pipeline components:", nlp.pipe_names)

doc = nlp("I love coffee.")

# Example 1: Lexical attributes
# spaCy computes various boolean flags (lexical attributes) for tokens.
print(f"{'Text':10} | {'is_alpha':10} | {'is_punct':10} | {'like_num':10}")
print("-" * 45)
for token in doc:
    print(f"{token.text:10} | {str(token.is_alpha):10} | {str(token.is_punct):10} | {str(token.like_num):10}")

# Example 2: Vocabulary
# spaCy stores data in a shared vocabulary (StringStore) to save memory.
# It hashes strings to integers.
coffee_hash = nlp.vocab.strings["coffee"]
coffee_string = nlp.vocab.strings[coffee_hash]
print(f"\nHash for 'coffee': {coffee_hash}")
print(f"String for hash {coffee_hash}: {coffee_string}")
```

## Phase 4: Tokenization
**Concepts**: Token objects, sentence segmentation, custom tokenizer rules

```python
# Phase 4: Tokenization
import spacy
from spacy.symbols import ORTH

nlp = spacy.load("en_core_web_sm")

# Example 1: Sentence Segmentation
text = "This is a sentence. This is another one! And a third?"
doc = nlp(text)
print("Sentences:")
for i, sent in enumerate(doc.sents):
    print(f"Sentence {i+1}: {sent.text}")

# Example 2: Exploring Token attributes
doc2 = nlp("Let's go to N.Y.!")
print("\nToken attributes:")
for token in doc2:
    print(f"{token.text:10} | Index: {token.i}")

# Example 3: Custom Tokenizer Rule
# Let's say we want to tokenize "gimme" as "gim" and "me"
special_case = [{ORTH: "gim"}, {ORTH: "me"}]
nlp.tokenizer.add_special_case("gimme", special_case)
doc3 = nlp("gimme that book.")
print("\nCustom Tokenization for 'gimme':", [t.text for t in doc3])
```

## Phase 5: Text Processing
**Concepts**: Lowercase, punctuation, stop words, filtering tokens

```python
# Phase 5: Text Processing
import spacy
nlp = spacy.load("en_core_web_sm")

text = "The quick brown fox jumps over the lazy dog! It's a sunny day."
doc = nlp(text)

# Example 1: Filtering out punctuation
no_punct = [token.text for token in doc if not token.is_punct]
print("Without punctuation:", no_punct)

# Example 2: Filtering out Stop Words (common words like 'the', 'is', 'a')
no_stop = [token.text for token in doc if not token.is_stop and not token.is_punct]
print("Without stop words & punctuation:", no_stop)

# Example 3: Lemmatization (base form of words)
# e.g., 'jumps' -> 'jump', 'is' -> 'be'
lemmas = [token.lemma_ for token in doc if not token.is_punct]
print("Lemmatized words:", lemmas)
```

## Phase 6: Linguistic Features
**Concepts**: POS tagging, morphology, dependency labels

```python
# Phase 6: Linguistic Features
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("She is running fast.")

# Example 1: Part-of-Speech Tagging and Dependency parsing
print(f"{'Text':10} | {'POS':6} | {'Tag':5} | {'Dep':10} | {'Head':10}")
print("-" * 50)
for token in doc:
    print(f"{token.text:10} | {token.pos_:6} | {token.tag_:5} | {token.dep_:10} | {token.head.text:10}")

# Example 2: Morphology
# Morphological features describe grammatical properties (e.g., singular/plural, tense).
print("\nMorphological Features:")
for token in doc:
    if str(token.morph):
        print(f"{token.text:10} -> {token.morph}")
```

## Phase 7: Part-of-Speech Tagging
**Concepts**: Noun, verb, adjective, adverb, tags and explanations

```python
# Phase 7: Part-of-Speech Tagging
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("The bright sun shines beautifully on the calm ocean.")

# Example 1: Extracting specific POS tags
adjectives = [token.text for token in doc if token.pos_ == "ADJ"]
verbs = [token.text for token in doc if token.pos_ == "VERB"]
adverbs = [token.text for token in doc if token.pos_ == "ADV"]

print("Adjectives:", adjectives)
print("Verbs:", verbs)
print("Adverbs:", adverbs)

# Example 2: Explaining POS Tags
# spaCy has a built-in explainer for its tags
print("\nTag Explanations:")
print("ADJ:", spacy.explain("ADJ"))
print("VERB:", spacy.explain("VERB"))
print("ADV:", spacy.explain("ADV"))
print("DET:", spacy.explain("DET"))
```

## Phase 8: Dependency Parsing
**Concepts**: Head words, dependency relations, syntax trees

```python
# Phase 8: Dependency Parsing
import spacy
from spacy import displacy # For visualization
nlp = spacy.load("en_core_web_sm")

doc = nlp("The cute cat chased the small mouse.")

# Example 1: Syntactic dependencies
print("Token dependencies:")
for token in doc:
    print(f"{token.text:10} is a {spacy.explain(token.dep_):30} attached to {token.head.text:10}")

# Example 2: Navigating the parse tree
print("\nNavigating the tree for 'chased':")
chased_token = doc[3] # 'chased'
print("Left children:", [t.text for t in chased_token.lefts])
print("Right children:", [t.text for t in chased_token.rights])

# Example 3: Noun Chunks
# Noun chunks are "base noun phrases" – flat phrases that have a noun as their head.
print("\nNoun Chunks:")
for chunk in doc.noun_chunks:
    print(f"{chunk.text:20} | Root: {chunk.root.text:10} | Dep: {chunk.root.dep_}")

# Note: In a real notebook, you can visualize the tree using:
# displacy.render(doc, style='dep', jupyter=True)
```

## Phase 9: Named Entity Recognition (NER)
**Concepts**: Persons, organizations, locations, dates, money, custom entities

```python
# Phase 9: Named Entity Recognition (NER)
import spacy
nlp = spacy.load("en_core_web_sm")

text = "Apple bought a UK startup for $1 billion on Tuesday."
doc = nlp(text)

# Example 1: Extracting entities
print(f"{'Entity':15} | {'Label':10} | {'Explanation'}")
print("-" * 50)
for ent in doc.ents:
    print(f"{ent.text:15} | {ent.label_:10} | {spacy.explain(ent.label_)}")

# Example 2: Accessing entities in spans
# We can find out if a token belongs to an entity
for token in doc:
    if token.ent_type_:
        print(f"Token '{token.text}' belongs to entity type: {token.ent_type_}")

# Note: You can visualize entities using:
# displacy.render(doc, style='ent', jupyter=True)
```

## Phase 10: Entity Ruler
**Concepts**: Rule-based entity extraction, patterns, matching

```python
# Phase 10: Entity Ruler
import spacy

nlp = spacy.load("en_core_web_sm")

# Example 1: Adding a custom rule-based entity recognizer
# We'll add the entity ruler before the statistical NER
ruler = nlp.add_pipe("entity_ruler", before="ner")

# Define patterns
patterns = [
    {"label": "TECH_COMPANY", "pattern": "TechNova"},
    {"label": "TECH_COMPANY", "pattern": [{"LOWER": "cyber"}, {"LOWER": "dynamics"}]}
]
ruler.add_patterns(patterns)

doc = nlp("TechNova is partnering with Cyber Dynamics to build AI.")

print("Extracted Entities with Custom Ruler:")
for ent in doc.ents:
    print(f"{ent.text:15} -> {ent.label_}")
```

## Phase 11: Matcher & PhraseMatcher
**Concepts**: Pattern matching, token rules, keyword extraction

```python
# Phase 11: Matcher & PhraseMatcher
import spacy
from spacy.matcher import Matcher, PhraseMatcher
nlp = spacy.load("en_core_web_sm")

# Example 1: Token-based Matcher
matcher = Matcher(nlp.vocab)
# Pattern: "learning" followed by "spaCy"
pattern1 = [{"LOWER": "learning"}, {"LOWER": "spacy"}]
# Pattern: Any adjective followed by "python"
pattern2 = [{"POS": "ADJ"}, {"LOWER": "python"}]

matcher.add("LEARNING_SPACY", [pattern1])
matcher.add("PYTHON_VIBES", [pattern2])

doc = nlp("I am learning spaCy. It is great for awesome python projects.")
matches = matcher(doc)

print("Matcher Results:")
for match_id, start, end in matches:
    span = doc[start:end]
    print(f"Rule: {nlp.vocab.strings[match_id]:15} | Match: {span.text}")

# Example 2: PhraseMatcher (Fast exact keyword matching)
phrase_matcher = PhraseMatcher(nlp.vocab)
terms = ["Barack Obama", "Angela Merkel", "Washington D.C."]
# Convert terms to Doc objects
patterns = [nlp.make_doc(text) for text in terms]
phrase_matcher.add("POLITICS", patterns)

doc2 = nlp("Barack Obama and Angela Merkel met in Washington D.C.")
matches2 = phrase_matcher(doc2)

print("\nPhraseMatcher Results:")
for match_id, start, end in matches2:
    span = doc2[start:end]
    print(span.text)
```

## Phase 12: Similarity & Word Vectors

```python
# Phase 12: Similarity & Word Vectors
# Note: 'en_core_web_sm' doesn't contain real word vectors (only context-sensitive tensors).
# To get good similarity results, you need a larger model like 'en_core_web_md' or 'en_core_web_lg'.
# For demonstration, we'll use 'sm', but results might not be perfect.

import spacy
nlp = spacy.load("en_core_web_sm")

# Example 1: Document Similarity
doc1 = nlp("I like apples and oranges.")
doc2 = nlp("I enjoy eating fruit.")
doc3 = nlp("The car is driving fast.")

print(f"Similarity (doc1 vs doc2): {doc1.similarity(doc2):.4f}")
print(f"Similarity (doc1 vs doc3): {doc1.similarity(doc3):.4f}")

# Example 2: Token Similarity
token1 = nlp("dog")[0]
token2 = nlp("cat")[0]
token3 = nlp("car")[0]

print(f"\nSimilarity (dog vs cat): {token1.similarity(token2):.4f}")
print(f"Similarity (dog vs car): {token1.similarity(token3):.4f}")

# Example 3: Word Vectors
print(f"\nVector for 'apple' (first 5 dims): {nlp('apple')[0].vector[:5]}")
```

## Phase 14: Sentiment Analysis
**Concepts**: Custom sentiment pipeline, integration with ML models

```python
# Phase 14: Sentiment Analysis
# spaCy doesn't have a built-in sentiment analyzer out of the box.
# However, you can add one using an extension attribute and a custom component.
# E.g., integrating TextBlob or VADER into spaCy.

import spacy
from spacy.tokens import Doc
# Using TextBlob for sentiment (uncomment to install: !pip install textblob)
from textblob import TextBlob

nlp = spacy.load("en_core_web_sm")

# Example 1: Registering a custom extension attribute
if not Doc.has_extension("sentiment"):
    Doc.set_extension("sentiment", default=0.0)

# Example 2: Creating a custom pipeline component
@spacy.Language.component("textblob_sentiment")
def textblob_sentiment(doc):
    # Calculate sentiment polarity (-1.0 to 1.0)
    sentiment = TextBlob(doc.text).sentiment.polarity
    doc._.sentiment = sentiment
    return doc

# Add component to pipeline
nlp.add_pipe("textblob_sentiment", last=True)

# Test it
texts = ["I love this beautiful world!", "This is the worst experience ever.", "It is okay, I guess."]

print(f"{'Text':40} | {'Sentiment Score'}")
print("-" * 60)
for text in texts:
    doc = nlp(text)
    print(f"{text:40} | {doc._.sentiment:.2f}")
```

## Phase 15: Training spaCy Models
**Concepts**: Training data, examples, annotations, optimization

```python
# Phase 15: Training spaCy Models
# In spaCy v3, training is done via config files and the CLI.
# Here, we demonstrate how to create training data in spaCy's `Example` format.

import spacy
from spacy.training import Example

nlp = spacy.blank("en")

# We want to train the model to recognize a new entity: "GADGET"
ner = nlp.add_pipe("ner")
ner.add_label("GADGET")

# Example 1: Creating training data
# Text and dictionary with entity spans
train_data = [
    ("I love my new iPhone 14.", {"entities": [(14, 23, "GADGET")]}),
    ("The Samsung Galaxy is a nice phone.", {"entities": [(4, 18, "GADGET")]})
]

# Example 2: Converting to spaCy Example objects
examples = []
for text, annotations in train_data:
    doc = nlp.make_doc(text)
    example = Example.from_dict(doc, annotations)
    examples.append(example)

print("Created Example objects for training.")
print("Example 1 Entities:", examples[0].reference.ents)

# (Actual training loop requires nlp.begin_training() and updating batches over epochs)
```

## Phase 16: Custom NER Training
**Concepts**: Label data, create entities, improve accuracy

```python
# Phase 16: Custom NER Training (Simulation)
import spacy
from spacy.training import Example
import random

nlp = spacy.blank("en")
ner = nlp.add_pipe("ner")
ner.add_label("ANIMAL")

# Training Data
train_data = [
    ("The dog barked loudly.", {"entities": [(4, 7, "ANIMAL")]}),
    ("I saw a cat in the garden.", {"entities": [(8, 11, "ANIMAL")]}),
    ("A wild bear appeared.", {"entities": [(7, 11, "ANIMAL")]})
]

# Convert to examples
examples = []
for text, ann in train_data:
    examples.append(Example.from_dict(nlp.make_doc(text), ann))

# Initialize training
optimizer = nlp.begin_training()

# Training loop (few epochs for demonstration)
print("Training started...")
for i in range(10):
    random.shuffle(examples)
    losses = {}
    nlp.update(examples, sgd=optimizer, losses=losses)
    if (i+1) % 5 == 0:
        print(f"Epoch {i+1} Losses:", losses)

# Testing the trained model
test_doc = nlp("My dog and my cat are friends.")
print("\nPredictions:")
for ent in test_doc.ents:
    print(f"{ent.text} -> {ent.label_}")
```

## Phase 17: Custom Components
**Concepts**: Create pipeline components, hooks, extensions

```python
# Phase 17: Custom Components
# We can create custom components to modify documents as they pass through the pipeline.

import spacy
from spacy.tokens import Token

nlp = spacy.load("en_core_web_sm")

# Example 1: Custom Token Extension
# Let's add a property to check if a word is a color
colors = ["red", "green", "blue", "yellow"]
Token.set_extension("is_color", getter=lambda token: token.text.lower() in colors, force=True)

# Example 2: Custom component that flags colors
@spacy.Language.component("color_flagger")
def color_flagger(doc):
    print("Executing color_flagger component...")
    for token in doc:
        if token._.is_color:
            print(f"  -> Found a color: {token.text}")
    return doc

# Add to pipeline
nlp.add_pipe("color_flagger", after="ner")

print("Pipeline:", nlp.pipe_names)

# Test it
doc = nlp("She wore a red dress and yellow shoes.")
```

## Phase 18: Data Annotation
**Concepts**: Prodigy, annotation formats, training datasets

```python
# Phase 18: Data Annotation
# Prodigy is a popular annotation tool by Explosion (creators of spaCy).
# Since Prodigy is a separate commercial tool, we'll demonstrate how to format data
# into spaCy's binary format (.spacy), which is required for training.

import spacy
from spacy.tokens import DocBin

nlp = spacy.blank("en")
db = DocBin()

# Raw data (simulating annotated data)
annotated_data = [
    ("Python is a great programming language.", [(0, 6, "PROGRAMMING_LANGUAGE")]),
    ("Java is also very popular.", [(0, 4, "PROGRAMMING_LANGUAGE")])
]

# Create Doc objects and add entities
for text, annotations in annotated_data:
    doc = nlp.make_doc(text)
    ents = []
    for start, end, label in annotations:
        span = doc.char_span(start, end, label=label)
        if span is not None:
            ents.append(span)
    doc.ents = ents
    db.add(doc)

# Save to disk
# db.to_disk("./train_data.spacy")
print(f"Successfully processed {len(db)} documents into DocBin.")
print("Ready to be saved as a .spacy file for CLI training.")
```

## Phase 19: Production Deployment
**Concepts**: Saving models, APIs, performance optimization

```python
# Phase 19: Production Deployment
import spacy

nlp = spacy.load("en_core_web_sm")

# Example 1: Saving a custom model to disk
# If we had trained a custom model, we could save it:
# nlp.to_disk("./my_custom_model")
print("Model can be saved using `nlp.to_disk('path')`")

# Example 2: Loading it back
# nlp_loaded = spacy.load("./my_custom_model")
print("Model can be loaded using `spacy.load('path')`")

# Example 3: Performance optimization with nlp.pipe
# When processing many texts, use `nlp.pipe` instead of calling `nlp()` in a loop.
# It processes texts in batches, which is much faster.

texts = ["Text one.", "Text two.", "Text three.", "Text four.", "Text five."]

# Inefficient:
# docs = [nlp(text) for text in texts]

# Efficient:
print("\nProcessing with nlp.pipe:")
docs = list(nlp.pipe(texts, batch_size=2))
for i, doc in enumerate(docs):
    print(f"Processed doc {i+1}: {doc.text}")
```
