```python
import nltk
from nltk.corpus import brown
from nltk import FreqDist

# Download the corpus (run once)
nltk.download("brown")

# Total number of words
print("Total words:", len(brown.words()))

# Available categories
print("Categories:")
print(brown.categories())

# First 20 words from the News category
print("\nFirst 20 words in News:")
print(brown.words(categories="news")[:20])

# Word frequency analysis
fdist = FreqDist(word.lower() for word in brown.words())

print("\nTop 10 most common words:")
for word, count in fdist.most_common(10):
    print(f"{word}: {count}")

# Count a specific word
word = "baby"
print(f"\n'{word}' appears {fdist[word]} times in the Brown Corpus.")
```

```python
import nltk
from nltk import word_tokenize, pos_tag

nltk.download("averaged_perceptron_tagger")
nltk.download("punkt")

text = "The quick brown fox jumps over the lazy dog."

tokens = word_tokenize(text)
tags = pos_tag(tokens)

print(tags)
```

```python
from nltk.corpus import wordnet as wn
import nltk

nltk.download("wordnet")

word = wn.synsets("car")[0]

print("Definition:", word.definition())
print("Synonyms:", word.lemma_names())

print("Hypernym:", word.hypernyms()[0].lemma_names())
print("Hyponyms:", word.hyponyms()[:3])

# Antonyms
good = wn.synsets("good")[0]
print("Antonym:", good.lemmas()[0].antonyms()[0].name())
```
